```python
import os
import random
from pathlib import Path

chunk_size = 1024 * 1024
num_sample = 10

input_file = Path("/home/shangfei/shang/cs336/assignment1-basics/data/TinyStoriesV2-GPT4-train.txt")
output_dir = Path("/home/shangfei/shang/cs336/assignment1-basics/data/sample/tiny")
output_dir.mkdir(exist_ok=True)

file_size = input_file.stat().st_size
starts = [random.randint(0, file_size-chunk_size) for _ in range(num_sample)]

with open(input_file, 'rb') as f:
    for i, start in enumerate(starts):
        f.seek(start)
        data = f.read(chunk_size)

        with open(output_dir / f"{i}.txt", 'w') as file:
            file.write(data.decode('utf-8', errors='replace'))
```

```python
import os
import random
from pathlib import Path

chunk_size = 1024 * 1024
num_sample = 10

input_file = Path("/home/shangfei/shang/cs336/assignment1-basics/data/owt_train.txt")
output_dir = Path("/home/shangfei/shang/cs336/assignment1-basics/data/sample/owt")
output_dir.mkdir(exist_ok=True)

file_size = input_file.stat().st_size
starts = [random.randint(0, file_size-chunk_size) for _ in range(num_sample)]

with open(input_file, 'rb') as f:
    for i, start in enumerate(starts):
        f.seek(start)
        data = f.read(chunk_size)

        with open(output_dir / f"{i}.txt", 'w') as file:
            file.write(data.decode('utf-8', errors='replace'))
```

```python
def get_compression_ratio(string: str, indices: list[int]) -> float:  
    """Given `string` that has been tokenized into `indices`, return the number of UTF-8 bytes per token.."""
    num_bytes = len(bytes(string, encoding="utf-8"))  
    num_tokens = len(indices)                       
    return num_bytes / num_tokens
```

```python
from tokenizer import Tokenizer
tiny_tokenizer = Tokenizer.from_files(
    vocab_filepath="/home/shangfei/shang/cs336/assignment1-basics/output/tiny_stores_vocab_train.pkl",
    merges_filepath="/home/shangfei/shang/cs336/assignment1-basics/output/tiny_stores_merges_train.pkl",
    special_tokens=["<|endoftext|>"]
)

owt_tokenizer = Tokenizer.from_files(
    vocab_filepath="/home/shangfei/shang/cs336/assignment1-basics/output/owt_vocab_train.pkl",
    merges_filepath="/home/shangfei/shang/cs336/assignment1-basics/output/owt_merges_train.pkl",
    special_tokens=["<|endoftext|>"]
)

tiny_folder = Path("/home/shangfei/shang/cs336/assignment1-basics/data/sample/tiny")
tiny_compression_ratio = 0
tiny_compression_ratio_owt = 0
for file in tiny_folder.iterdir():
    with open(file, 'r') as f:
        data = f.read()
        indices = tiny_tokenizer.encode(data)
        indices_owt = owt_tokenizer.encode(data)
        tiny_compression_ratio += get_compression_ratio(data, indices)
        tiny_compression_ratio_owt += get_compression_ratio(data, indices_owt)

owt_folder = Path("/home/shangfei/shang/cs336/assignment1-basics/data/sample/owt")
owt_compression_ratio = 0
owt_compression_ratio_tiny = 0
for file in owt_folder.iterdir():
    with open(file, 'r') as f:
        data = f.read()
        indices = owt_tokenizer.encode(data)
        indices_tiny = tiny_tokenizer.encode(data)
        owt_compression_ratio += get_compression_ratio(data, indices)
        owt_compression_ratio_tiny += get_compression_ratio(data, indices_tiny)
```
